# Week 1, Day 1: Data Science Primer 

### What is data science? 
- Multi-disciplinary field which aims to derive **insights** and **knowledge** from both structured and unstructured **data**

### Why the sudden interest? 
#### Big Data
- Because of internet and connectivity, we generate so much data now!
**V's of Big Data**
**Volume**
  - refers to the huge amount of data that is generated daily
**Variety**
  - refers to the diversity of *data types* and *data sources*
**Velocity**
  - refers to the *speed* at which data is generated, processed and analyzed
**Veracity**
  - refers to the authenticity and credibility of the data (and in turn, its quality)
**Value**
  - refers to the added value for various stakeholder (company, government, everyone!)

Computing power became easier to access and setting up a supercomputer that is "pay-per-use" 

#### Level of Analytics: Low to High
- Descriptive -> Diagnostic -> Predictive -> Prescriptive

### Types of Analysis 
#### Descriptive 
**Descriptive**
- Aggregating totals - What is happening?
- How many COVID cases going upwards or downwards
#### Diagnostic 
**Exploratory** 
- Spread and central tendency - What is the average? Are there outliers? Trend?
- Is the trend of COVID cases going upwards or downwards? 
**Inferential**
- Difference between two samples
- Correlation is part of inference
- Is the daily change of cases in NCR the same as the change of cases in Region 4? 
#### Predictive 
**Causal**
- Establish and estimate relationship of two variables
- If mask wearing was not implementd in the Philippines, how will it affect the number of cases in the country? 
**Predictive**
- Looking beyond available data to see what will happen in the future
- How many COVID cases will happen in the next month? 
#### Prescriptive 
**Mechanistic**
- Effect estimation through simulation of variables (Policy making)
- How does mask wearing reduce the transmission of COVID?
